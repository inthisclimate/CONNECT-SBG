```python
#repeat the image's GPS and quadrat/transect info 3 times and merge with SED data.
#this is because we took 3 spectral SED samples per quadrat

quadrat_repeated = quadrat_df.loc[
    quadrat_df.index.repeat(3)
].reset_index(drop=True)

quadrat_repeated["replicate"] = (
    quadrat_repeated.groupby("quadrat_id").cumcount() + 1
)

print(quadrat_repeated.shape)
print(quadrat_repeated.head(12))
```

```python
# 4. ASSIGN EACH SED MEASUREMENT TO A QUADRAT

# Reset the index so both tables run from row 0 to row 176
spectra_metadata = spectra_metadata.reset_index(drop=True)

# Combine the SED filename with the repeated quadrat information
spectral_quadrat_metadata = pd.concat(
    [
        spectra_metadata[["filename"]],
        quadrat_repeated
    ],
    axis=1 #put the tables beside each other as new columns, rather than stacking one underneath the other.
)

print(spectral_quadrat_metadata.shape)

spectral_quadrat_metadata.head(12)
```

```python
# 5. CHECK TRANSECT TRANSITIONS

# Show the end of Transect 7 and beginning of Transect 8
spectral_quadrat_metadata[
    spectral_quadrat_metadata["quadrat_id"].between(40, 43)
]
```

```python
# Check the end of Transect 9 and beginning of Transect 10
# we should not see a quadrat 54 in here, which was not sampled. 
spectral_quadrat_metadata[
    spectral_quadrat_metadata["quadrat_id"].between(52, 56)
]
```

```python
# 6. VALIDATE SPECTRAL-QUADRAT ASSIGNMENTS

# Check total numbers
assert len(quadrat_df) == 59, "Expected 59 sampled quadrats."
assert len(spectra_metadata) == 177, "Expected 177 SED measurements."
assert len(spectral_quadrat_metadata) == 177, "Expected 177 assigned spectra."

# Check that every sampled quadrat has exactly 3 spectra
spectra_per_quadrat = (
    spectral_quadrat_metadata
    .groupby("quadrat_id")
    .size()
)

assert (spectra_per_quadrat == 3).all(), \
    "At least one quadrat does not have exactly 3 spectra."

# Check first and last SED measurements
assert spectral_quadrat_metadata.iloc[0]["filename"] == "1594452_00097"
assert spectral_quadrat_metadata.iloc[-1]["filename"] == "1594452_00273"

# Check that Quadrat 54 is correctly absent
assert 54 not in spectral_quadrat_metadata["quadrat_id"].values

print("All validation checks passed.")
print(f"Quadrats: {spectral_quadrat_metadata['quadrat_id'].nunique()}")
print(f"Spectra: {len(spectral_quadrat_metadata)}")
print(f"Transects: {spectral_quadrat_metadata['transect'].nunique()}")
```

```python
# 7. SAVE SPECTRAL-QUADRAT METADATA

spectral_quadrat_metadata.to_csv(
    output_file,
    index=False
)

print(f"Saved quadrat assignments to:\n{output_file}")
```
